# Demo 7: Deep RL met CartPole

Een tabel heeft één rij per situatie nodig. De toestand van CartPole bestaat uit vier doorlopende getallen, dus er zijn eindeloos veel situaties. Een klein neuraal netwerk neemt de plaats van de tabel in.

**Zo gebruik je het:** klik op *Run → Run All Cells*. Wil je iets veranderen, pas dan de cel met instellingen aan en voer die cel en de cel eronder opnieuw uit.

```python
from rl_demos.cartpole import network_diagram

network_diagram()
```

```python
# Settings: change these and run this cell again
GAMES = 600              # stop after this many games at most
DISCOUNT = 0.99          # how much future rewards count
LEARNING_RATE = 0.001    # how big each adjustment to the network is
HIDDEN_NEURONS = 64      # neurons per hidden layer
USE_PRETRAINED = False   # True loads a trained network (only with HIDDEN_NEURONS = 64)
SEED = 3                 # None gives a different run each time
```

```python
from rl_demos.cartpole import CartPoleDemo, before_after

demo = CartPoleDemo(games=GAMES, discount=DISCOUNT, learning_rate=LEARNING_RATE,
                    hidden=HIDDEN_NEURONS, use_pretrained=USE_PRETRAINED, seed=SEED)
demo
```

Druk op **Train**. Hij bewaart het beste netwerk dat hij tot nu toe heeft gezien en stopt zodra dat netwerk de volle 500 stappen in balans blijft. Voer daarna de cel hieronder uit (bij *Run All* draait die vóór het trainen, dus voer hem daarna nog een keer uit).
