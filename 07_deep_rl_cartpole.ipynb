# Demo 7: Deep RL with CartPole

A table needs one row per situation. CartPole's state is four continuous numbers, so there are endless situations. A small neural network replaces the table.

**How to use:** click *Run → Run All Cells*. To change something, edit the settings cell, then run it and the cell below it again.

```python
from rl_demos.cartpole import network_diagram

network_diagram()
```

```python
# Settings: change these and run this cell again
GAMES = 600              # stop after this many games at most
DISCOUNT = 0.99          # how much future rewards count
LEARNING_RATE = 0.001    # how big each adjustment to the network is
HIDDEN_NEURONS = 64      # neurons per hidden layer
USE_PRETRAINED = False   # True loads a trained network (only with HIDDEN_NEURONS = 64)
SEED = 3                 # None gives a different run each time
```

```python
from rl_demos.cartpole import CartPoleDemo, before_after

demo = CartPoleDemo(games=GAMES, discount=DISCOUNT, learning_rate=LEARNING_RATE,
                    hidden=HIDDEN_NEURONS, use_pretrained=USE_PRETRAINED, seed=SEED)
demo
```

Press **Train**. It keeps the best network it has seen and stops when that one balances for the full 500 steps. Then run the cell below (with *Run All*, it runs before training, so run it again afterwards).

```python
before_after(demo.network)
```
