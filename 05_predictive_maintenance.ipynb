## Het probleem

Een machine verdient geld op elke dag dat hij draait, en slijt op sommige dagen een beetje (ongeveer één dag op de drie) (slijtage 0 = nieuw, 9 = versleten). Hoe meer hij versleten is, hoe groter de kans dat hij vandaag kapotgaat.

Elke dag ziet de robot het slijtageniveau en kiest een van twee dingen:

- **niets doen:** blijven draaien en de omzet van de dag verdienen
- **repareren:** een dag stilstaan, een reparatie betalen, en de machine is weer zo goed als nieuw

Gaat de machine kapot, dan is er geen keuze meer: hij moet **vervangen** worden, en dat is veel duurder. De vaste regel is "draaien tot hij kapotgaat".

Druk op **Train** en daarna op **Play a year**. Kijk naar de slijtagebalk en naar de policy-tabel: de robot leert een *drempel*, "niets doen tot slijtage X, dan repareren".

```python
# Settings: change these and run this cell again
DAILY_REVENUE = 10        # earned each day the machine runs
REPAIR_COST = 30          # cost of a planned repair (the machine is as good as new)
REPLACE_COST = 200        # cost of replacing a broken machine
TRAINING_DAYS = 300_000   # simulated days of practice
USE_PRETRAINED = False    # True skips training (only valid with the default costs and probabilities)
SEED = 0
```

```python
# Breakdown probabilities: the chance that the machine breaks today, for each wear level
# Use values from 0 to 1 (0.05 = 5%), then run this cell and the two below again
BREAKDOWN_PROBABILITIES = [
    0.0,      # wear 0 (new)
    0.00048,  # wear 1
    0.00384,  # wear 2
    0.01296,  # wear 3
    0.03073,  # wear 4
    0.06001,  # wear 5
    0.10370,  # wear 6
    0.16468,  # wear 7
    0.24582,  # wear 8
    0.35,     # wear 9 (worn out)
]
```

```python
from rl_demos.applied import MaintenanceDemo

maintenance_demo = MaintenanceDemo(revenue=DAILY_REVENUE, repair_cost=REPAIR_COST, replace_cost=REPLACE_COST,
                                   breakdown_chances=BREAKDOWN_PROBABILITIES,
                                   training_days=TRAINING_DAYS, use_pretrained=USE_PRETRAINED, seed=SEED)
maintenance_demo
```

```python
# The Q-matrix: one row per wear level, one column per action
# It updates while you train. Run this cell again whenever you run the demo cell again.
from rl_demos.qmatrix import show_maintenance_q

show_maintenance_q(maintenance_demo)
```
