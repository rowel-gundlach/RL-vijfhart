# Demo 5: Predictive maintenance

The same Q-learning, now on a business problem: when do you repair a machine? It is compared with a simple fixed rule.

**How to use:** click *Run → Run All Cells*. To change something, edit a settings cell, then run it and the cells below it again.

## The problem

A machine earns money every day it runs, and wears a little most days (wear 0 = new, 9 = worn out). The more worn it is, the more likely it breaks down today.

Each day the robot sees the wear level and picks one of two things:

- **do nothing:** keep running and earn the day's revenue
- **repair:** stop for the day, pay for a repair, and the machine is as good as new

If the machine breaks, there is no choice left: it must be **replaced**, which is much more expensive. The fixed rule is "run until it breaks".

Press **Train**, then **Play a year**. Watch the wear bar, and look at the policy table: the robot learns a *threshold*, "do nothing until wear X, then repair".

```python
# Settings: change these and run this cell again
DAILY_REVENUE = 10        # earned each day the machine runs
REPAIR_COST = 30          # cost of a planned repair (the machine is as good as new)
REPLACE_COST = 200        # cost of replacing a broken machine
TRAINING_DAYS = 300_000   # simulated days of practice
USE_PRETRAINED = False    # True skips training (only valid with the default costs and probabilities)
SEED = 0
```

```python
# Breakdown probabilities: the chance that the machine breaks today, for each wear level
# Use values from 0 to 1 (0.05 = 5%), then run this cell and the two below again
BREAKDOWN_PROBABILITIES = [
    0.0,      # wear 0 (new)
    0.00048,  # wear 1
    0.00384,  # wear 2
    0.01296,  # wear 3
    0.03073,  # wear 4
    0.06001,  # wear 5
    0.10370,  # wear 6
    0.16468,  # wear 7
    0.24582,  # wear 8
    0.35,     # wear 9 (worn out)
]
```

```python
from rl_demos.applied import MaintenanceDemo

maintenance_demo = MaintenanceDemo(revenue=DAILY_REVENUE, repair_cost=REPAIR_COST, replace_cost=REPLACE_COST,
                                   breakdown_chances=BREAKDOWN_PROBABILITIES,
                                   training_days=TRAINING_DAYS, use_pretrained=USE_PRETRAINED, seed=SEED)
maintenance_demo
```

```python
# The Q-matrix: one row per wear level, one column per action
# It updates while you train. Run this cell again whenever you run the demo cell again.
from rl_demos.qmatrix import show_maintenance_q

show_maintenance_q(maintenance_demo)
```
