# Demo 1: Train je eigen hond

De hond begrijpt geen taal. Hij heeft alleen een tabel met getallen: één rij per woord, één kolom per kunstje. Een snoepje duwt een getal omhoog, een "Nee!" duwt het omlaag.

**Zo gebruik je het:** klik op *Run → Run All Cells*. Wil je iets veranderen, pas dan de cel met instellingen aan en voer die cel en de cel eronder opnieuw uit.

```python
# Starting settings (exploration and learning rate also have sliders in the demo)
EXPLORATION = 0.15     # how often the dog tries something random (0 = never, 1 = always)
LEARNING_RATE = 0.5    # how far one treat moves the number (0 to 1)
TREAT_REWARD = 1       # reward for a treat
NO_REWARD = -1         # reward for a "No!"
SEED = None            # a number gives the same dog every time
```

```python
from rl_demos.dog import DogDemo

DogDemo(exploration=EXPLORATION, learning_rate=LEARNING_RATE, treat_reward=TREAT_REWARD,
        no_reward=NO_REWARD, seed=SEED)
```
