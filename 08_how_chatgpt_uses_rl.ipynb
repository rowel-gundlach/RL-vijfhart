## Part B: Why does ChatGPT agree with you so easily?

Under every ChatGPT answer there is a 👍 and a 👎. Those ratings are used to retrain the model: answers that get a 👍 become more likely. Here **you** are the person who rates.

A tiny network has learned from text that the capital of the Netherlands is Amsterdam, but it is not completely sure. It sees two things, and you choose both:

- **what the user claims**: *Amsterdam* (right) or *The Hague* (wrong, it is the seat of government),
- **how the user says it**: *asks* or *insists*.

1. Choose *The Hague* and *insists*, then press **Ask the model**.
2. Rate the answer: 👍 or 👎. Be a typical user: you like hearing that you are right.
3. Ask and rate a few times. Watch the lines in the network and the table below it. Did your ratings also change the situations you never rated?
4. Press **Back to: trained on text**, then **Let thousands of people rate**: simulated people who mostly reward correct answers, and like being agreed with a little.

Orange pulls towards "you're right", blue towards a correction.

*This is a toy: a real model has no separate "what I know" input, and the people here are simulated. The pressure from the ratings is the same kind.*

```python
# Settings for part B: what do the simulated people give a 👍 for? (used by "Let thousands of people rate")
LIKES_TRUTH = 0.5       # extra chance of a 👍 for a correct answer
LIKES_AGREEMENT = 0.2   # extra chance of a 👍 for agreeing with the user
```

```python
from rl_demos.rlhf import SycophancyDemo

SycophancyDemo(likes_truth=LIKES_TRUTH, likes_agreement=LIKES_AGREEMENT)
```
