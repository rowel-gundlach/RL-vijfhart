## Deel B: Waarom is ChatGPT het zo makkelijk met je eens?

Onder elk antwoord van ChatGPT staan een 👍 en een 👎. Met die beoordelingen wordt het model opnieuw getraind: antwoorden die een 👍 krijgen, worden waarschijnlijker. Hier ben **jij** degene die beoordeelt.

Een piepklein netwerk heeft uit tekst geleerd dat Amsterdam de hoofdstad van Nederland is, en het is daar vrij zeker van: tegen *Den Haag* zegt het bijna altijd "nee". Het ziet twee dingen, en jij kiest ze allebei:

- **wat de gebruiker beweert**: *Amsterdam* (goed) of *Den Haag* (fout, daar zit de regering),
- **hoe de gebruiker het zegt**: *vraagt* of *houdt vol*.

1. Kies *Den Haag* en *houdt vol* en druk op **Vraag het model**.
2. Beoordeel het antwoord: 👍 of 👎. Wees een gewone gebruiker: je hoort graag dat je gelijk hebt.
3. Vraag en beoordeel een keer of tien. Kijk naar de lijnen in het netwerk en de tabel eronder. Zijn door jouw beoordelingen ook de situaties veranderd die je nooit hebt beoordeeld?
4. Druk op **Terug naar: getraind op tekst** en dan op **Laat duizenden mensen beoordelen**: gesimuleerde mensen die vooral juiste antwoorden belonen, en het een beetje fijn vinden als je het met ze eens bent.

Oranje trekt richting "je hebt gelijk", blauw richting een correctie.

*Dit is speelgoed: een echt model heeft geen aparte ingang voor "wat ik weet", en de mensen hier zijn gesimuleerd. De druk van de beoordelingen is wel van dezelfde soort.*

```python
# Settings for part B: what do the simulated people give a 👍 for? (used by "Let thousands of people rate")
LIKES_TRUTH = 0.5       # extra chance of a 👍 for a correct answer
LIKES_AGREEMENT = 0.2   # extra chance of a 👍 for agreeing with the user
```

```python
from rl_demos.rlhf import SycophancyDemo

SycophancyDemo(likes_truth=LIKES_TRUTH, likes_agreement=LIKES_AGREEMENT)
```
