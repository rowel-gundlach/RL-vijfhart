# Demo 2: Snake, rules vs. learning

Hand-written rules work until the world surprises them. Learning from rewards finds its own rules.

**How to use:** click *Run → Run All Cells*. To change something, edit the settings cell, then run it and the code cells below it again.

```python
# Settings: change these and run this cell again
APPLE_REWARD = 10        # reward for eating an apple
DEATH_REWARD = -10       # reward for crashing
STEP_REWARD = -0.1       # small cost per move, so the snake does not wander
LEARNING_RATE = 0.1      # how far one experience moves a number in the table
DISCOUNT = 0.9           # how much future rewards count (0 to 1)
GAMES = 10_000           # how many games to train for
SHOW_EVERY = 2500        # show one game every this many games (each one takes about 10 seconds to watch)
USE_PRETRAINED = False   # True starts with an expert snake (fallback if training is slow)
SEED = 0                 # None gives a different run each time
```

```python
from rl_demos.snake import SnakeDemo

snake = SnakeDemo(apple_reward=APPLE_REWARD, death_reward=DEATH_REWARD, step_reward=STEP_REWARD,
                  learning_rate=LEARNING_RATE, discount=DISCOUNT, games=GAMES, show_every=SHOW_EVERY,
                  use_pretrained=USE_PRETRAINED, seed=SEED)
snake
```
