# Demo 4: Dynamic pricing

The same Q-learning, on a second business problem: what should a ticket cost today? It is compared with a simple fixed rule.

**How to use:** click *Run → Run All Cells*, type a number of concerts next to **Train** (try 100 first, then 60000) and press it. Pressing again trains further. To change something, edit a settings cell, then run it and the cells below it again.

## The problem

Selling concert tickets. Each day the seller sees the state, picks a price, and chance decides what happens next.

| RL term | Here | Where you set it |
|---|---|---|
| **State** | days left + tickets left | `TICKETS`, `TICKET_GROUPS`, one row per day in `BUY_PROBABILITIES` |
| **Action** | today's price | `PRICES` |
| **Reward** | price × tickets sold, plus the value of unsold tickets at the end | `PRICES`, `UNSOLD_TICKET_VALUE` |
| **Transition probabilities** | how many tickets are left tomorrow | `CUSTOMERS_PER_DAY`, `BUY_PROBABILITIES` |

The transition: every day some people buy a ticket. How many depends on the price and on the day: with the default numbers people pay more as the concert gets closer. **Tomorrow's tickets = today's tickets − today's buyers.** The fixed rule is the best single price, found by trying them all.

When you press **Sell the concert**, both sellers sell side by side, day by day, with the same luck. You see each day's price, how many people bought, the seats filling up and the money coming in.

```python
# The problem: change these and run this cell again
PRICES = [50, 75, 100, 125, 150]   # the actions: prices to choose from, low to high
TICKETS = 200                      # tickets available at the start
TICKET_GROUPS = 20                 # the state: tickets left are seen in this many groups
CUSTOMERS_PER_DAY = 30             # how many people look at the tickets each day (on average)
UNSOLD_TICKET_VALUE = 0            # reward for each empty seat at the end (-20 = an empty seat costs €20)
```

```python
# Transition probabilities: the chance that someone who looks buys, for each day and each price
# One row per day: add or remove rows to change the days of selling. One column per price in PRICES.
# Use values from 0 to 1 (0.05 = 5%), then run this cell and the cells below again
BUY_PROBABILITIES = [
    # €50   €75    €100   €125   €150
    [0.788, 0.421, 0.115, 0.014, 0.001],  # 14 days left (first day of selling)
    [0.792, 0.426, 0.118, 0.014, 0.001],  # 13
    [0.804, 0.443, 0.126, 0.016, 0.001],  # 12
    [0.823, 0.471, 0.142, 0.019, 0.001],  # 11
    [0.848, 0.511, 0.165, 0.024, 0.001],  # 10
    [0.876, 0.562, 0.199, 0.033, 0.002],  # 9
    [0.905, 0.622, 0.245, 0.046, 0.004],  # 8
    [0.933, 0.690, 0.307, 0.066, 0.006],  # 7
    [0.956, 0.761, 0.385, 0.098, 0.011],  # 6
    [0.974, 0.829, 0.480, 0.147, 0.020],  # 5
    [0.987, 0.889, 0.587, 0.218, 0.038],  # 4
    [0.994, 0.936, 0.698, 0.315, 0.069],  # 3
    [0.998, 0.967, 0.801, 0.438, 0.124],  # 2
    [0.999, 0.986, 0.885, 0.579, 0.212],  # 1 day left
]
```

```python
# How the seller learns: change these and run this cell again
LEARNING_RATE = 0.1   # how far one concert moves a number in the Q-matrix
EXPLORATION = 0.4     # how often it tries a random price at first (it explores less as it learns)
DISCOUNT = 1.0        # how much the coming days count (1 = as much as today)
SEED = 0
```

```python
from rl_demos.applied import PricingDemo

pricing_demo = PricingDemo(prices=PRICES, tickets=TICKETS, ticket_groups=TICKET_GROUPS,
                           customers_per_day=CUSTOMERS_PER_DAY, unsold_value=UNSOLD_TICKET_VALUE,
                           buy_chances=BUY_PROBABILITIES,
                           learning_rate=LEARNING_RATE, exploration=EXPLORATION, discount=DISCOUNT, seed=SEED)
pricing_demo
```

```python
# The Q-matrix on the final day (1 day left): tickets left × price
# It updates while you train. Run this cell again whenever you run the demo cell again.
from rl_demos.qmatrix import show_pricing_q

show_pricing_q(pricing_demo)
```
